# List Manipulation and Operations

manipulate a list

```python
buah = ['apple','mangga','jeruk','anggur','pisang']
print(f'{'list contents' :<40}: {buah}')

print('=' * 120)

buah.append("semangka")
print(f"{'add watermelon to the list' :<40}: {buah}\n")

buah.insert(3,"durian")
print(f"{'add durian among the oranges and grapes' :<40}: {buah}\n")

buah.remove("mangga")
print(f"{'Remove the mango from the list' :<40}: {buah}\n")

buah[4] = "nanas"
print(f"{'turn a banana into a pineapple' :<40}: {buah}\n")

print(f"{'Display the first 3 items in the list' :<40}: {buah[0],buah[1],buah[2]}")
```

sort a list

```python
angka = [45, 12, 78, 23, 56, 89, 34]
print(f'{'list content' :<32}: {angka}')

print('=' * 70)

angka.sort()
print(f"{'Sort the list in ascending order' :<32}: {angka}\n")

angka.reverse()
print(f"{'Sort a list in descending order' :<32}: {angka}")
```
