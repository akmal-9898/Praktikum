### - tuple
What did i learn?
1. Creating a tuple
2. Changing one of the values in a tuple

```python
data = ("zidane arrizqi akmal", 2505060054,"teknologi informasi")
print(f'{'tuple' :<6}: {data}\n')

print('modify data by adding additional data in the same sequence')
update = data[2] + " dan elektro"
print(f'{'prodi' :<6}: {update}\n')

print('display one piece of data from a tuple')
print(f'{'NIM' :<6}: {data[1]}')
```

### - Dictionary
What did i learn?
1. create a dictionary
2. use a loop to count integer data types
3. display data in a specific format

```python
data = {
    "apel": 5000,
    "pisang": 3000,
    "jeruk": 7000,
    "anggur": 15000
}
print(f'{'dictionary' :<12}: {data}\n')

total = 0
for price in data.values():
  total += price
print(f'{'total(IDR)' :<12}: {total}\n')

print('display the contents of a dictionary in a specific format')
for key,values in data.items():
  print(f"{'Harga'} {key :<6}: Rp {values}")
```
